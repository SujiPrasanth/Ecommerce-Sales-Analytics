```python
df['Weekend_Order']=df["Order_Date"].dt.dayofweek.map( lambda x: "Weekend" if x >=5 else "Weekday")
```

```python
df.head()
```

```python
def age_grp(age):
    if age<20:
        return "Teens"
    elif age<=30:
        return 'Young Adult'
    elif age<=40:
        return 'Adult'
    elif age<=50:
        return "Middle Age"
    else:
        return "Seniors"
```

```python
df['Age_Group']=df['Customer_Age'].apply(age_grp)
```
